```python
# PASO 3: Instalar las dependencias del proyecto
# Assignment 5 - Reproducibility

from pathlib import Path
import subprocess
import sys

# 1. Localizar el proyecto descargado desde GitHub
carpetas = list(
    Path("/content").glob("*/Assignments/Assignment5_Reproducibility")
)

if not carpetas:
    raise FileNotFoundError(
        "No se encontró la carpeta Assignment5_Reproducibility."
    )

proyecto = carpetas[0]
requirements = proyecto / "requirements.txt"

# 2. Verificar que exista requirements.txt
if not requirements.is_file():
    raise FileNotFoundError(
        "No se encontró el archivo requirements.txt."
    )

print("Proyecto encontrado:", proyecto)
print("\nDependencias que se instalarán:")
print(requirements.read_text(encoding="utf-8"))

# 3. Instalar las versiones indicadas en requirements.txt
print("\nInstalando dependencias...")

subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-r",
        str(requirements)
    ],
    check=True
)

print("\nINSTALACIÓN COMPLETADA CORRECTAMENTE.")
```

```python
# PASO 4: Ejecutar el experimento y comprobar su reproducibilidad
# Assignment 5 - Reproducibility

from pathlib import Path
import subprocess
import sys
import pandas as pd

# 1. Encontrar la carpeta del proyecto descargado
carpetas = list(
    Path("/content").glob("*/Assignments/Assignment5_Reproducibility")
)

if not carpetas:
    raise FileNotFoundError("No se encontró la carpeta del proyecto.")

proyecto = carpetas[0]
archivo_entrenamiento = proyecto / "train.py"
archivo_resultados = proyecto / "results.csv"

if not archivo_entrenamiento.is_file():
    raise FileNotFoundError("No se encontró train.py.")

print("PROYECTO:", proyecto)
print("\nEjecutando entrenamiento desde GitHub...\n")

# 2. Ejecutar train.py
ejecucion = subprocess.run(
    [sys.executable, str(archivo_entrenamiento)],
    cwd=str(proyecto),
    capture_output=True,
    text=True
)

print(ejecucion.stdout)

if ejecucion.returncode != 0:
    print("ERROR DURANTE LA EJECUCIÓN:")
    print(ejecucion.stderr)
    raise RuntimeError("El entrenamiento no terminó correctamente.")

# 3. Leer los resultados generados
resultados = pd.read_csv(archivo_resultados)

media = resultados["accuracy"].mean()
desviacion = resultados["accuracy"].std()

# 4. Comparar con el experimento original
print("\n========== VERIFICACIÓN ==========")
print(f"Accuracy promedio obtenido: {media:.4f}")
print(f"Desviación estándar: {desviacion:.4f}")

if round(media, 4) == 0.9842 and round(desviacion, 4) == 0.0073:
    print("\nPRUEBA EXITOSA: se reprodujeron los resultados originales.")
else:
    print("\nLos resultados son diferentes. Debemos investigar la causa.")
```
