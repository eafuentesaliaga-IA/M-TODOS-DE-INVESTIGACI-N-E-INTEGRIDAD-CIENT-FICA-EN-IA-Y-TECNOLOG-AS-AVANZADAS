```python
def render_card(card, title="Model Card"):
    lines = [f"# {title}", ""]
    for section, fields in card.items():
        lines.append(f"## {section}")
        for k, v in fields.items():
            lines.append(f"- **{k}:** {v}")
        lines.append("")
    return "\n".join(lines)

card_md = render_card(model_card)
print(card_md)

# Save it so you can submit the rendered Markdown
with open("model_card.md", "w", encoding="utf-8") as f:
    f.write(card_md)
print("\n\u2705 Saved to model_card.md  (File > Download, or right-click in the file panel)")
```
